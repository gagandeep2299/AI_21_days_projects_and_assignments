#### *Choose a diffusion model and load the pipeline*

```python
import torch
from diffusers import DiffusionPipeline
import matplotlib.pyplot as plt
```

```python
pipeline_test2image = DiffusionPipeline.from_pretrained("stabilityai/stable-diffusion-xl-base-1.0", torch_dtype=torch.float16, use_safetensors=True, variant="fp16")
pipeline_test2image.to('cuda')
```

```python
prompt = "a school boy boarding the bus with dog,4k,cinematic lighting"
generated_image = pipeline_test2image(prompt=prompt).images[0]
```

```python
plt.figure(figsize=(10,10))
plt.imshow(generated_image)
plt.axis("off")
plt.title(prompt)
plt.show()
```
